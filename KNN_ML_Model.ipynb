 KNN Model — Telecom Customer Churn Prediction

Objective: Build a KNN classification model to predict whether a telecom customer will churn (leave) or stay.

Dataset: Telecom customer data including usage patterns, service calls, and plan details.

Algorithm: K-Nearest Neighbors (KNN) — a simple distance-based classification algorithm.

Tools: Python · NumPy · Pandas · Seaborn · Matplotlib · Scikit-learn

1. Import Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

2.Import Dataset

```python
data = pd.read_csv('https://raw.githubusercontent.com/DataAnalyst21/DatasetsForDataAnalytics/refs/heads/main/telecom_churn_clean.csv')
data.info()
```

3.EDA (Exploratory Data Analysis)

```python
# general overview
data.describe()
```

```python
# churn balance
data['churn'].value_counts().plot(kind='pie', autopct='%1.1f%%',
                                   colors=['#3498db','#e74c3c'])
plt.title('Churn vs Stay')
plt.ylabel('')
plt.show()
```

```python
# find best predictors
plt.figure(figsize=(13, 8))
sns.heatmap(data.corr(numeric_only=True), annot=True,
            fmt='.2f', cmap='RdYlGn', center=0)
plt.title('Feature Correlation')
plt.show()
```

4. Preparation for Model

```python
# selected features based on correlation results
features = data[['total_day_minutes', 'voice_mail_plan',
                  'total_eve_charge']].values
target = data['churn']
```

```python
# 75% train, 25% test
features_train, features_test, target_train, target_test = train_test_split(
    features, target, test_size=0.25, random_state=10
)
```

```python
# scale the features
sc = StandardScaler()
features_train = sc.fit_transform(features_train)
features_test  = sc.transform(features_test)
```

5. Fit the Model

```python
# build and train
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(features_train, target_train)

accuracy = knn.score(features_test, target_test)
print(f'Accuracy = {accuracy * 100:.2f}%')
```

6. Predict

```python
# predict a new customer
# [total_day_minutes, number_customer_service_calls, total_eve_charge]
new_customer = np.array([[230, 3, 20]])
new_customer = sc.transform(new_customer)

result = knn.predict(new_customer)
print('Customer will Churn!' if result == 1 else 'Customer will Stay!')
```
